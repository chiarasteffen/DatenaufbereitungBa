# Aufbereitung der analysierten Daten

```python
# Imports
from pathlib import Path
import pandas as pd
import numpy as np

# Pfad zum Data-Ordner
data_dir = Path("Data")
```

```python
# Excel aus Notebook 2 einlesen
df = pd.read_excel(data_dir / "02_API_Baujahr_Adressen.xlsx").copy()

print("Shape:", df.shape)
print("\nSpalten:")
print(df.columns.tolist())

display(df.head())
```

```python
# Grundlegende Datenübersicht visualisieren

print("Datentypen:")
display(df.dtypes.to_frame("dtype"))

print("\nFehlende Werte pro Spalte:")
display(df.isna().sum().sort_values(ascending=False).to_frame("fehlende_werte"))

print("\nAnzahl eindeutige Werte pro Spalte:")
display(df.nunique(dropna=False).sort_values(ascending=False).to_frame("anzahl_eindeutige_werte"))

# Erste deskriptive Übersicht für numerische Spalten
print("\nDeskriptive Statistik (numerische Spalten):")
display(df.describe(include="number").T)

# Deskriptive Übersicht für Text-/Kategorie-Spalten
print("\nDeskriptive Statistik (Objekt-/Kategorie-Spalten):")
display(df.describe(include="object").T)
```

```python
# Baujahr aus der API als führende Baujahr-Spalte verwenden
df["BAUJAHR_API"] = pd.to_numeric(df["baujahr_gwr"], errors="coerce")

# Baujahr-Kategorien bilden
df["BAUJAHR_KATEGORIE"] = pd.cut(
    df["BAUJAHR_API"],
    bins=[0, 1800, 1850, 1900, 1945, 1960, 1980, 2000, 2025],
    labels=[
        "bis 1800",
        "1801-1850",
        "1851-1900",
        "1901-1945",
        "1946-1960",
        "1961-1980",
        "1981-2000",
        "2001-2024"
    ],
    include_lowest=True
)

print("Verteilung Baujahr-Kategorien:")
display(df["BAUJAHR_KATEGORIE"].value_counts(dropna=False).sort_index())

print("\nBeispielansicht:")
display(df[["EGID", "STRASSENNAME", "HAUSNR", "ORT", "BAUJAHR_API", "BAUJAHR_KATEGORIE"]].head(20))
```

```python
# Auf Gebäudeebene wechseln: pro EGID genau ein Gebäude behalten
df_buildings = (
    df.sort_values(["EGID", "STRASSENNAME", "HAUSNR"])
      .drop_duplicates(subset="EGID", keep="first")
      .reset_index(drop=True)
)

print("Anzahl Zeilen vorher:", len(df))
print("Anzahl eindeutige Gebäude nach EGID:", len(df_buildings))

# Häufigkeiten der wichtigsten Merkmale
merkmale = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster"
]

for merkmal in merkmale:
    print(f"\n=== Häufigkeit: {merkmal} ===")
    display(
        df_buildings[merkmal]
        .value_counts(dropna=False)
        .to_frame("Anzahl")
    )
```

```python
# Kategorien für die spätere Stichprobe verdichten

df_prep = df_buildings.copy()

# Fenster bereinigen
df_prep["Fenster_KATEGORIE"] = df_prep["Fenster"].copy()
df_prep["Fenster_KATEGORIE"] = df_prep["Fenster_KATEGORIE"].fillna("Unbekannt")
df_prep["Fenster_KATEGORIE"] = df_prep["Fenster_KATEGORIE"].replace({
    "nein": "KEIN",
    "Nein": "KEIN"
})

# Nutzung auf Top-Kategorien reduzieren
top_nutzung = df_prep["NUTZUNG"].value_counts().head(10).index
df_prep["NUTZUNG_KATEGORIE"] = df_prep["NUTZUNG"].where(
    df_prep["NUTZUNG"].isin(top_nutzung),
    "Sonstige"
)

# Eigentumskategorie auf wichtige Kategorien reduzieren
top_eigentum = df_prep["GS_EIGENTUMSKATEGORIE"].value_counts().head(5).index
df_prep["GS_EIGENTUMSKATEGORIE_KATEGORIE"] = df_prep["GS_EIGENTUMSKATEGORIE"].where(
    df_prep["GS_EIGENTUMSKATEGORIE"].isin(top_eigentum),
    "Sonstige"
)

# Kontrolle der verdichteten Kategorien
merkmale_reduziert = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE_KATEGORIE",
    "Fenster_KATEGORIE"
]

for merkmal in merkmale_reduziert:
    print(f"\n=== Verdichtete Häufigkeit: {merkmal} ===")
    display(
        df_prep[merkmal]
        .value_counts(dropna=False)
        .to_frame("Anzahl")
    )
```

```python
# Basis für Ground-Truth-Datensatz aufbauen

df_gt_basis = df_buildings.copy()

# 1) Nur die wichtigsten Hauptnutzungen behalten
wichtige_hauptnutzung = [
    "Verwaltungsgebäude und Gebäude mit öffentlichem Charakter",
    "Wohngebäude",
    "Industrie und Gerwerbe"
]

df_gt_basis = df_gt_basis[df_gt_basis["HAUPTNUTZUNG"].isin(wichtige_hauptnutzung)].copy()

# 2) Nur die häufigsten Eigentumskategorien behalten
wichtige_eigentum = [
    "Immobilien",
    "Schulamt",
    "Übrige",
    "Tiefbau",
    "Sportamt"
]

df_gt_basis = df_gt_basis[df_gt_basis["GS_EIGENTUMSKATEGORIE"].isin(wichtige_eigentum)].copy()

# 3) Fenster bereinigen: nur klare Kategorien behalten
df_gt_basis["Fenster_KATEGORIE"] = df_gt_basis["Fenster"].fillna("Unbekannt").replace({
    "nein": "KEIN",
    "Nein": "KEIN"
})

df_gt_basis = df_gt_basis[df_gt_basis["Fenster_KATEGORIE"].isin(["Bevor 1990", "Ab 1990", "KEIN"])].copy()

# 4) Nutzung auf Top-10 reduzieren
top_nutzung_10 = df_gt_basis["NUTZUNG"].value_counts().head(10).index
df_gt_basis["NUTZUNG_KATEGORIE"] = df_gt_basis["NUTZUNG"].where(
    df_gt_basis["NUTZUNG"].isin(top_nutzung_10),
    "Sonstige"
)

# 5) Übersicht der reduzierten Basis
print("Anzahl Gebäude in Ground-Truth-Basis:", len(df_gt_basis))

for merkmal in [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]:
    print(f"\n=== {merkmal} ===")
    display(df_gt_basis[merkmal].value_counts(dropna=False).to_frame("Anzahl"))
```

```python
# Vorschlag für repräsentativen Ground-Truth-Datensatz (max. 15 Gebäude)

df_gt = df_gt_basis.copy()

# 1) Seltenheitsscore berechnen:
# seltene Kategorien bekommen einen höheren Score
score_spalten = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]

for spalte in score_spalten:
    haeufigkeit = df_gt[spalte].value_counts(normalize=True)
    df_gt[f"score_{spalte}"] = df_gt[spalte].map(lambda x: 1 / haeufigkeit[x])

df_gt["DIVERSITAETS_SCORE"] = df_gt[[f"score_{s}" for s in score_spalten]].sum(axis=1)

# 2) Auswahl aufbauen
auswahl = []

# 2a) Jede Baujahr-Kategorie einmal abdecken
for kategorie in df_gt["BAUJAHR_KATEGORIE"].dropna().value_counts().index:
    kandidat = (
        df_gt[
            (df_gt["BAUJAHR_KATEGORIE"] == kategorie) &
            (~df_gt["EGID"].isin([x["EGID"] for x in auswahl]))
        ]
        .sort_values("DIVERSITAETS_SCORE", ascending=False)
        .head(1)
    )
    if not kandidat.empty:
        auswahl.append(kandidat.iloc[0].to_dict())

# 2b) Jede Hauptnutzung mindestens einmal abdecken
for kategorie in df_gt["HAUPTNUTZUNG"].dropna().value_counts().index:
    kandidat = (
        df_gt[
            (df_gt["HAUPTNUTZUNG"] == kategorie) &
            (~df_gt["EGID"].isin([x["EGID"] for x in auswahl]))
        ]
        .sort_values("DIVERSITAETS_SCORE", ascending=False)
        .head(1)
    )
    if not kandidat.empty:
        auswahl.append(kandidat.iloc[0].to_dict())

# 2c) Rest bis max. 15 mit höchstem Diversitätsscore auffüllen
bereits_gewaehlt = [x["EGID"] for x in auswahl]

rest = (
    df_gt[~df_gt["EGID"].isin(bereits_gewaehlt)]
    .sort_values("DIVERSITAETS_SCORE", ascending=False)
)

noch_noetig = 15 - len(auswahl)

if noch_noetig > 0:
    for _, row in rest.head(noch_noetig).iterrows():
        auswahl.append(row.to_dict())

# 3) Ergebnis als DataFrame
df_ground_truth = pd.DataFrame(auswahl).drop_duplicates(subset="EGID").reset_index(drop=True)

# Falls durch Überschneidungen weniger als 15 vorhanden sind, ist das okay
print("Anzahl Ground-Truth-Gebäude:", len(df_ground_truth))

display(df_ground_truth[[
    "EGID",
    "STRASSENNAME",
    "HAUSNR",
    "PLZ4",
    "ORT",
    "lat_api",
    "lon_api",
    "BAUJAHR_API",
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE",
    "DIVERSITAETS_SCORE"
]].sort_values(
    ["BAUJAHR_KATEGORIE", "HAUPTNUTZUNG", "DIVERSITAETS_SCORE"],
    ascending=[True, True, False]
))
```

```python
# Repräsentativität der 15 Ground-Truth-Gebäude prüfen
# und Score-Berechnung robuster machen

df_gt_check = df_gt_basis.copy()

score_spalten = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]

# Robuste Score-Berechnung: nur tatsächlich vorkommende Kategorien verwenden
for spalte in score_spalten:
    haeufigkeit = df_gt_check[spalte].astype(str).value_counts(normalize=True)
    df_gt_check[f"score_{spalte}"] = df_gt_check[spalte].astype(str).map(
        lambda x: 1 / haeufigkeit[x] if haeufigkeit[x] > 0 else 0
    )

df_gt_check["DIVERSITAETS_SCORE"] = df_gt_check[[f"score_{s}" for s in score_spalten]].sum(axis=1)

# Die bereits ausgewählten EGIDs mit der Basis verknüpfen
df_ground_truth_check = df_gt_check[df_gt_check["EGID"].isin(df_ground_truth["EGID"])].copy()

# Vergleich der Verteilungen: Gesamtbasis vs. Ground-Truth
vergleich_spalten = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]

for spalte in vergleich_spalten:
    print(f"\n=== Vergleich: {spalte} ===")
    
    gesamt = (
        df_gt_check[spalte]
        .astype(str)
        .value_counts(normalize=True)
        .rename("Gesamtanteil")
    )
    
    gt = (
        df_ground_truth_check[spalte]
        .astype(str)
        .value_counts(normalize=True)
        .rename("GT_Anteil")
    )
    
    vergleich = pd.concat([gesamt, gt], axis=1).fillna(0)
    vergleich["Differenz"] = vergleich["GT_Anteil"] - vergleich["Gesamtanteil"]
    
    display(vergleich.sort_values("Gesamtanteil", ascending=False))
```

```python
# Neue, repräsentativere Auswahl mit proportionalen Quoten
df_basis = df_gt_basis.copy()

# Hilfsfunktion: proportionale Zielanzahlen mit Largest-Remainder-Methode
def allocate_counts(series, total_n):
    anteile = series.value_counts(normalize=True)
    raw = anteile * total_n
    basis = np.floor(raw).astype(int)
    rest = raw - basis

    fehlend = total_n - basis.sum()
    if fehlend > 0:
        for idx in rest.sort_values(ascending=False).index[:fehlend]:
            basis[idx] += 1

    return basis[basis > 0]

# 1) Quoten nach HAUPTNUTZUNG für insgesamt 15 Gebäude
ziel_anzahl = 15
quoten_hauptnutzung = allocate_counts(df_basis["HAUPTNUTZUNG"], ziel_anzahl)

print("Quoten HAUPTNUTZUNG:")
display(quoten_hauptnutzung.to_frame("Zielanzahl"))

auswahl_rows = []
bereits_gewaehlt = set()

# 2) Innerhalb jeder Hauptnutzung proportional nach BAUJAHR_KATEGORIE sampeln
for hauptnutzung, n_haupt in quoten_hauptnutzung.items():
    teilmenge_h = df_basis[df_basis["HAUPTNUTZUNG"] == hauptnutzung].copy()

    quoten_baujahr = allocate_counts(teilmenge_h["BAUJAHR_KATEGORIE"], n_haupt)

    for baujahr_kat, n_baujahr in quoten_baujahr.items():
        kandidaten = teilmenge_h[
            (teilmenge_h["BAUJAHR_KATEGORIE"] == baujahr_kat) &
            (~teilmenge_h["EGID"].isin(bereits_gewaehlt))
        ].copy()

        if len(kandidaten) == 0:
            continue

        sample_n = min(n_baujahr, len(kandidaten))
        sample = kandidaten.sample(n=sample_n, random_state=42)

        auswahl_rows.append(sample)
        bereits_gewaehlt.update(sample["EGID"].tolist())

# 3) Falls wir wegen knapper Kombinationen noch unter 15 sind: aus Rest auffüllen
df_ground_truth_v2 = pd.concat(auswahl_rows, ignore_index=True) if auswahl_rows else pd.DataFrame()

if len(df_ground_truth_v2) < ziel_anzahl:
    rest = df_basis[~df_basis["EGID"].isin(df_ground_truth_v2["EGID"])].copy()
    noch_noetig = ziel_anzahl - len(df_ground_truth_v2)

    if len(rest) > 0:
        extra = rest.sample(n=min(noch_noetig, len(rest)), random_state=42)
        df_ground_truth_v2 = pd.concat([df_ground_truth_v2, extra], ignore_index=True)

# 4) Sicherheitshalber deduplizieren
df_ground_truth_v2 = df_ground_truth_v2.drop_duplicates(subset="EGID").reset_index(drop=True)

print("Anzahl Ground-Truth-Gebäude (v2):", len(df_ground_truth_v2))

display(df_ground_truth_v2[[
    "EGID",
    "STRASSENNAME",
    "HAUSNR",
    "PLZ4",
    "ORT",
    "lat_api",
    "lon_api",
    "BAUJAHR_API",
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]].sort_values(["HAUPTNUTZUNG", "BAUJAHR_API"]))
```

```python
# Repräsentativität der neuen Auswahl (v2) prüfen

vergleich_spalten = [
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]

for spalte in vergleich_spalten:
    print(f"\n=== Vergleich v2: {spalte} ===")

    gesamt = (
        df_gt_basis[spalte]
        .astype(str)
        .value_counts(normalize=True)
        .rename("Gesamtanteil")
    )

    gt = (
        df_ground_truth_v2[spalte]
        .astype(str)
        .value_counts(normalize=True)
        .rename("GT_Anteil")
    )

    vergleich = pd.concat([gesamt, gt], axis=1).fillna(0)
    vergleich["Differenz"] = vergleich["GT_Anteil"] - vergleich["Gesamtanteil"]

    display(vergleich.sort_values("Gesamtanteil", ascending=False))
```

```python
from pathlib import Path

# Finale Ground-Truth-Tabelle aufbereiten
df_ground_truth_final = df_ground_truth_v2.copy()

# Vollständige Adresse ergänzen
df_ground_truth_final["ADRESSE"] = (
    df_ground_truth_final["STRASSENNAME"].astype(str).str.strip()
    + " "
    + df_ground_truth_final["HAUSNR"].astype(str).str.strip()
    + ", "
    + df_ground_truth_final["ORT"].astype(str).str.strip()
)

# Relevante Spalten auswählen und sortieren
df_ground_truth_final = df_ground_truth_final[[
    "EGID",
    "ADRESSE",
    "STRASSENNAME",
    "HAUSNR",
    "PLZ4",
    "ORT",
    "lat_api",
    "lon_api",
    "BAUJAHR_API",
    "BAUJAHR_KATEGORIE",
    "HAUPTNUTZUNG",
    "NUTZUNG_KATEGORIE",
    "GS_EIGENTUMSKATEGORIE",
    "Fenster_KATEGORIE"
]].sort_values(
    ["HAUPTNUTZUNG", "BAUJAHR_API", "ADRESSE"]
).reset_index(drop=True)

print("Finale Anzahl Ground-Truth-Gebäude:", len(df_ground_truth_final))
display(df_ground_truth_final)
```

```python
# Export in den Data-Ordner
data_dir = Path("Data")
data_dir.mkdir(exist_ok=True)

output_path = data_dir / "03_Ground_Truth_Datensatz.xlsx"
df_ground_truth_final.to_excel(output_path, index=False)

print(f"Gespeichert: {output_path}")
```
