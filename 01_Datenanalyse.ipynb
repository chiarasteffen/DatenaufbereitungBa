# Datenaufbereitung für Ground Trouth Datensatz

```python
# Imports 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Einstellungen für die Darstellung
data_dir = Path("Data")
data_dir.mkdir(exist_ok=True)

# Pakages
%pip install pandas openpyxl numpy matplotlib seaborn
```

```python
# Daten einlesen
df = pd.read_excel(data_dir / "Adressen_Winti_Dep_Arch.xlsx").copy()
df.head()
```

```python
# Daten anzeigen
print("Shape:", df.shape)
print("\nSpalten:")
print(df.columns.tolist())

print("\nInfo:")
df.info()

print("\nFehlende Werte pro Spalte:")
print(df.isna().sum())

if "EGID" in df.columns:
    print("\nAnzahl eindeutige EGIDs:", df["EGID"].nunique())
    print("Anzahl Zeilen:", len(df))
    print("Anzahl doppelte EGIDs:", len(df) - df["EGID"].nunique())
```

```python
# Mögliche Baujahr-Spalten suchen
baujahr_kandidaten = [c for c in df.columns if "bau" in c.lower() or "jahr" in c.lower()]
print("Mögliche Baujahr-Spalten:", baujahr_kandidaten)
```

```python
# Alle unterschiedlichen Werte aus der Spalte BAUJAHR einmalig anzeigen
baujahr_werte = (
    df["BAUJAHR"]
    .dropna()                  # leere Werte entfernen
    .astype(str)               # alles einheitlich als Text behandeln
    .str.strip()               # Leerzeichen entfernen
    .replace("", pd.NA)        # leere Strings als NA behandeln
    .dropna()                  # erneut leere Werte entfernen
    .unique()                  # nur eindeutige Werte
)

# sortiert ausgeben
baujahr_werte = sorted(baujahr_werte)

print("Eindeutige Werte in BAUJAHR:")
for wert in baujahr_werte:
    print(wert)
```

```python
# Zeilen anzeigen, in denen BAUJAHR = 0, 1000 oder 1200 ist

werte = [0, 1000, 1200]

df_baujahr_filter = df[df["BAUJAHR"].isin(werte)].copy()

print(df_baujahr_filter)
```

```python
# Finale Tabelle aus Notebook 01 exportieren
output_path = data_dir / "01_Datenanalyse_Ergebnis.xlsx"
df.to_excel(output_path, index=False)

print(f"Gespeichert: {output_path}")
print("Anzahl Zeilen:", len(df))
print("Spalten:", df.columns.tolist())
```
