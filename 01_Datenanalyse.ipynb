```python
# Imports 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Pakages
!pip install pandas openpyxl numpy matplotlib seaborn
```

```python
# Daten einlesen

df = pd.read_excel("Adressen_Winti_Dep_Arch.xlsx")
df.head()
```

```python
# Daten anzeigen

print(df.shape)
print(df.columns)
df.info()
```

```python
# Spaltennamen anzeigen
print(df.columns)
```

```python
# Spaltennamen 'EGID', 'GEB_GEBID', 'STRASSENNAME', 'HAUSNR', 'HAUSNRZUSATZ', 'PLZ4', 'ORT', 'STADTKREIS', 'HAUPTNUTZUNG' auswählen und in neuem df speichern
df_selected = df[['EGID', 'GEB_GEBID', 'STRASSENNAME', 'HAUSNR', 'HAUSNRZUSATZ', 'PLZ4', 'ORT', 'STADTKREIS', 'HAUPTNUTZUNG']]

# df_selected anzeigen
df_selected.head()
```

```python
# df_selected Spalte Hauptnutzung analysieren und Häufigkeit der Werte anzeigen
print(df_selected['HAUPTNUTZUNG'].value_counts())
```

```python
# Neues df aus df_selected erstellen, nur mit den Zeilen, in denen Hauptnutzung == 'Verwaltungsgebäude und Gebäude mit öffentlichem Charakter' oder 'Wohngebäude' ist
df_filtered = df_selected[df_selected['HAUPTNUTZUNG'].isin(['Verwaltungsgebäude und Gebäude mit öffentlichem Charakter', 'Wohngebäude'])]

# df_filtered anzeigen
print(df_filtered.head())

# Anzahl der Zeilen in df_filtered anzeigen
print(df_filtered.shape[0])
```

```python
# 6 Zeilen mit Hauptnutzung 'Verwaltungsgebäude und Gebäude mit öffentlichem Charakter' und 6 Zeilen mit Hauptnutzung 'Wohngebäude' in neuem df speichern
df_final = df_filtered[df_filtered['HAUPTNUTZUNG'].isin(['Verwaltungsgebäude und Gebäude mit öffentlichem Charakter', 'Wohngebäude'])].groupby('HAUPTNUTZUNG').head(6)

# df_final anzeigen
print(df_final)
```

```python
# df_final als Excel-Datei speichern
df_final.to_excel("Datenanalyse_Ergebnis.xlsx", index=False)
```
