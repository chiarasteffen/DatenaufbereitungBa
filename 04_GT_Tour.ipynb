# Dieses Notebook wird den Ground Trouth Datensatz so aufbereiten, sodass eine Tour dafür berechnet werden kann

```python
# Imports
import pandas as pd
from pathlib import Path

# Pfad zum Data-Ordner
data_dir = Path("Data")
```

```python
# Excel-Datei aus Notebook 03 einlesen
df = pd.read_excel(data_dir / "03_Ground_Truth_Datensatz.xlsx").copy()

print("Shape:", df.shape)
print("\nSpalten:")
print(df.columns.tolist())

display(df.head())
```

```python
# Neues DataFrame für die Tour erstellen
df_tour = pd.DataFrame({
    "Nr": [""] * len(df),
    "Adresse": df["STRASSENNAME"].astype(str).str.strip() + " "
               + df["HAUSNR"].astype(str).str.strip() + ", "
               + df["PLZ4"].astype(str).str.strip() + " "
               + df["ORT"].astype(str).str.strip(),
    "lat": df["lat_api"],
    "lon": df["lon_api"]
})

# Überprüfen der neuen Struktur
print("Shape:", df_tour.shape)
print("\nSpalten:")
print(df_tour.columns.tolist())
display(df_tour.head())
```

```python
# Tour-Datensatz exportieren
output_path = data_dir / "04_GT_Tour.xlsx"
df_tour.to_excel(output_path, index=False)
print(f"Gespeichert: {output_path}")
```
